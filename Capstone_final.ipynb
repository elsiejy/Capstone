## 3. Data Cleansing & Preparation

Let's clean the dataset. We'll identify missing values, handle empty rows, correct data types, and prepare the dataset for analysis.

Data cleansing on booking dataset:
1. Derive check-in date
2. Update data type for reservation_status_date
3. Update data type for is_repeated_guest and is_canceled
4. Update data type for children, agent and company
5. Handle NULL values for children, country, agent, company

```python
# 1. Derive check-in date
# combine arrival_date_year, arrival_date_month and arrival_date_day_of_month into strings
booking_date_strings = (
    bookings['arrival_date_year'].astype(str) + '-' +
    bookings['arrival_date_month'].astype(str) + '-' +
    bookings['arrival_date_day_of_month'].astype(str)
)

# convert the strings into a proper pandas datetime object
bookings['checkin_date'] = pd.to_datetime(booking_date_strings)

# verify that first 5 rows of checkin_date
print(bookings['checkin_date'].head())
```

```python
# 2. Update data type for reservation_status_date
# convert reservation_status_date to datetime format
bookings['reservation_status_date'] = pd.to_datetime(bookings['reservation_status_date'])

# view the first 5 rows of reservation_status_date
print(bookings['reservation_status_date'].head())
```

```python
# 3. Update data type for is_repeated_guest and is_canceled
# change datatype from int to boolean
bookings[['is_repeated_guest', 'is_canceled']] = bookings[['is_repeated_guest', 'is_canceled']].astype(bool)

# verify count of is_repeated_guest and is_canceled
print(bookings['is_canceled'].value_counts())
print(bookings['is_repeated_guest'].value_counts())

# confirm values are only false, true
print(bookings['is_canceled'].unique())
print(bookings['is_repeated_guest'].unique())
```

```python
# 4. Update data type for children, agent and company
# change datatype from float to int
bookings[['children', 'agent', 'company']] = bookings[['children', 'agent', 'company']].astype('Int64')
```

```python
# 5. Handle NULL values for children, country, agent, company
# handle NULL value for children by replacing NULL with 0
bookings['children'] = bookings['children'].fillna(0)

# handle NULL value for country by replacing NULL with UNKNOWN
bookings['country'] = bookings['country'].fillna('Unknown')

# handle NULL value for agent by replacing NULL with 0
bookings['agent'] = bookings['agent'].fillna(0)

# handle NULL value for company by replacing NULL with 0
bookings['company'] = bookings['company'].fillna(0)
```

### 0.1 Data check

`lead_time` counts **backwards** from arrival, so `booking_date = checkin_date − lead_time`. `reservation_status_date` is the date the *last* status was set, so its meaning depends on the status:

* **Check-Out** → the checkout date (≈ check-in + total nights)
* **Canceled** → the cancellation date
* **No-Show** → the arrival date

Example — Booking 9: booked 7 Apr 2015 (1 Jul − 85 days), cancelled 6 May 2015, due to arrive 1 Jul 2015. It was cancelled 29 days after booking and **56 days before arrival**. The check below confirms the rule holds across the whole file.

```python
gap = (df['reservation_status_date'] - df['checkin_date']).dt.days
co = df['reservation_status'].eq('Check-Out')
ca = df['reservation_status'].eq('Canceled')
ns = df['reservation_status'].eq('No-Show')

checks = pd.DataFrame({
    'rows': [co.sum(), ca.sum(), ns.sum()],
    'rule': ['status date = check-in + nights', 'cancelled on/before arrival and on/after booking',
             'status date = check-in'],
    'rows_meeting_rule': [(gap[co] == df.loc[co, 'total_nights']).sum(),
                          ((gap[ca] <= 0) & (df.loc[ca, 'reservation_status_date'] >= df.loc[ca, 'booking_date'])).sum(),
                          (gap[ns] == 0).sum()],
}, index=['Check-Out', 'Canceled', 'No-Show'])
checks['% meeting rule'] = (checks['rows_meeting_rule'] / checks['rows']).map('{:.2%}'.format)
display(checks)
print(df.loc[df['Booking_ID'] == 9, ['booking_date', 'reservation_status_date', 'checkin_date',
                                    'reservation_status', 'days_before_cancel']].to_string(index=False))
```

The dates are consistent: 99.96% of check-outs leave on check-in + nights (the 27 exceptions are same-day departures), every cancellation is dated between booking and arrival, and no-shows are recorded on the arrival date. `reservation_status_date` can therefore be used as the **cancellation date**

```python
def summarise(d, by):
    # Volume, cancellation and revenue metrics for any grouping.
    t = d.groupby(by, observed=True).agg(
        bookings=('is_canceled', 'size'), cancellations=('is_canceled', 'sum'),
        booked_rev=('room_revenue', 'sum'), rev_at_risk=('rev_at_risk', 'sum'),
        late_rev_at_risk=('late_rev_at_risk', 'sum'), median_adr=('adr', 'median'),
        avg_nights=('total_nights', 'mean'))
    t['cancel_rate'] = t['cancellations'] / t['bookings']
    t['rev_loss_rate'] = t['rev_at_risk'] / t['booked_rev']
    t['share_bookings'] = t['bookings'] / len(d)
    t['share_cancellations'] = t['cancellations'] / d['is_canceled'].sum()
    t['share_rev_at_risk'] = t['rev_at_risk'] / d['rev_at_risk'].sum()
    t['late_share_of_risk'] = t['late_rev_at_risk'] / t['rev_at_risk']
    return t

def show(t, cols=None, sort=None, n=None):
    # Display a summary table with readable %, $ and count formats.
    t = t.copy()
    if sort: t = t.sort_values(sort, ascending=False)
    if n: t = t.head(n)
    cols = cols or list(t.columns)
    out = pd.DataFrame(index=t.index)
    for c in cols:
        v = t[c]
        if c.endswith('_rate') or c.startswith('share_') or c.endswith('_share') or c.endswith('_of_risk') or c.startswith('pct'):
            out[c] = v.map(lambda x: f'{x:.1%}' if pd.notna(x) else '')
        elif (c.startswith('rev') or c.endswith('_rev') or 'rev_at_risk' in c) and not c.startswith('rank'):
            out[c] = v.map(lambda x: f'${x/1e6:,.2f}M' if abs(x) >= 1e5 else f'${x:,.0f}')
        elif c == 'median_adr':
            out[c] = v.map(lambda x: f'${x:,.0f}')
        elif not pd.api.types.is_numeric_dtype(v):
            out[c] = v
        elif pd.api.types.is_float_dtype(v):
            out[c] = v.map(lambda x: f'{x:,.1f}')
        else:
            out[c] = v.map(lambda x: f'{x:,}')
    display(out)

def three_panel(t, title, figsize=(12, 3.2), order=None, overall=None):
    # Bookings | cancellation rate | revenue at risk, as small multiples on one category axis.
    t = t.loc[order] if order is not None else t.sort_values('rev_at_risk')
    labels = [str(i) for i in t.index]
    fig, axes = plt.subplots(1, 3, figsize=figsize, sharey=True)
    specs = [('bookings', NEUTRAL, 'Bookings', lambda x: f'{x:,.0f}'),
             ('cancel_rate', CANC, 'Cancellation rate', lambda x: f'{x:.0%}'),
             ('rev_at_risk', CANC, 'Revenue at risk', lambda x: f'${x/1e6:,.1f}M')]
    for ax, (col, colr, name, f) in zip(axes, specs):
        ax.barh(labels, t[col], color=colr, height=0.62)
        for y, v in enumerate(t[col]):
            ax.text(v, y, ' ' + f(v), va='center', fontsize=8.5, color=INK2)
        ax.set_title(name, fontsize=10)
        ax.grid(axis='y', visible=False)
        ax.margins(x=0.22)
        ax.xaxis.set_major_locator(mtick.MaxNLocator(4))
        if col == 'cancel_rate':
            ax.xaxis.set_major_formatter(PCT)
            if overall is not None:
                ax.axvline(overall, color=INK2, ls='--', lw=0.8)
        if col == 'rev_at_risk': ax.xaxis.set_major_formatter(EUR_M)
    fig.suptitle(title, x=0.01, ha='left', fontweight='bold', fontsize=12)
    fig.tight_layout()
    plt.show()

def cramers_v(x, y):
    ct = pd.crosstab(x, y)
    chi2 = chi2_contingency(ct, correction=False)[0]
    n = ct.values.sum(); r, k = ct.shape
    return np.sqrt(chi2 / (n * (min(r, k) - 1)))

OVERALL_RATE = df['is_canceled'].mean()
OVERALL_LOSS = df['rev_at_risk'].sum() / df['room_revenue'].sum()
```

```python
# 2.3 Strength of association — Cramér's V for every candidate characteristic
def cramers_v(x, y):
    ct = pd.crosstab(x, y)
    chi2 = chi2_contingency(ct, correction=False)[0]
    n = ct.values.sum(); r, k = ct.shape
    return np.sqrt(chi2 / (n * (min(r, k) - 1)))

cands = {'Deposit type': 'deposit_type', 'Lead time band': 'lead_band', 'Market segment': 'market_segment',
         'Special requests': 'requests_band', 'Distribution channel': 'distribution_channel',
         'Customer type': 'customer_type', 'Booking changes': 'changes_band', 'Hotel type': 'hotel',
         'Previous cancellation': 'prev_cancel_flag', 'Length of stay': 'stay_band', 'Party composition': 'party',
         'Waiting list band': 'wait_band', 'Repeat guest': 'is_repeated_guest', 'Meal plan': 'meal',
         'Arrival month': 'arrival_mname', 'Has agent': 'has_agent',
         'Car parking requested*': 'required_car_parking_spaces', 'Country = PRT*': None}
rows = []
for label, col in cands.items():
    col = col or 'country_prt'
    x = df[col] if col != 'required_car_parking_spaces' else (df[col] > 0).astype(int)
    rows.append((label, cramers_v(x, df['is_canceled'])))
cv = pd.DataFrame(rows, columns=['feature', 'cramers_v']).sort_values('cramers_v')
fig, ax = plt.subplots(figsize=(9, 6.6))
cols = [CONTEXT if f.endswith('*') else BLUE for f in cv['feature']]
ax.barh(cv['feature'], cv['cramers_v'], color=cols, height=0.62)
for yi, v in enumerate(cv['cramers_v']):
    ax.text(v, yi, f' {v:.2f}', va='center', fontsize=8.5)
ax.set_xlim(0, cv['cramers_v'].max() * 1.15); ax.grid(axis='y', visible=False); ax.set_xlabel("Cramér's V (0 = none, 1 = perfect)")
suptitle(fig, '2.3  Deposit type, lead time and market segment are the strongest single signals',
         "Association with cancellation (Cramér's V). * grey = recorded at/after arrival, excluded from the model")
fig.subplots_adjust(left=0.25); plt.show()
```

Deposit type has the strongest unadjusted association with cancellation in this set (Cramér's V = 0.481).

```python
# 3.3 Logistic regression — odds ratio for each waiting-list band, unadjusted vs adjusted
import statsmodels.formula.api as smf
mdl = seg.copy()
mdl['wait_band'] = mdl['wait_band'].astype(str)
for c in ['lead_band', 'stay_band', 'party', 'requests_band', 'arrival_mname']:
    mdl[c] = mdl[c].astype(str)
ref = "C(wait_band, Treatment('0'))"
f_un = f"is_canceled ~ {ref}"
f_adj = (f"is_canceled ~ {ref} + C(hotel) + C(market_segment) + C(deposit_type) + C(customer_type) "
         "+ C(lead_band) + C(stay_band) + C(party) + C(requests_band) + C(arrival_mname) "
         "+ is_repeated_guest + prev_cancel_flag")
res = {}
for name, f in [('Unadjusted', f_un), ('Adjusted', f_adj)]:
    fit = smf.logit(f, data=mdl).fit(disp=0, maxiter=300)
    ci = fit.conf_int()
    keep = [i for i in fit.params.index if i.startswith(ref)]
    res[name] = pd.DataFrame({'OR': np.exp(fit.params[keep]).values, 'lo': np.exp(ci.loc[keep, 0]).values,
                              'hi': np.exp(ci.loc[keep, 1]).values, 'p': fit.pvalues[keep].values},
                             index=[k.split('[T.')[1].rstrip(']') for k in keep]).reindex(WAIT[1:])
fig, ax = plt.subplots(figsize=(9.5, 4))
y = np.arange(len(WAIT) - 1)
for k, (name, c) in enumerate([('Unadjusted', CONTEXT), ('Adjusted', BLUE)]):
    r = res[name]; yy = y + (0.15 if k == 0 else -0.15)
    ax.errorbar(r['OR'], yy, xerr=[r['OR'] - r['lo'], r['hi'] - r['OR']], fmt='o', color=c, ms=7,
                capsize=0, elinewidth=2, label=name, markeredgecolor=SURFACE)
ax.axvline(1, color=INK2, lw=1)
ax.set_xscale('log'); ax.xaxis.set_major_formatter(mtick.FuncFormatter(lambda v, _: f'{v:g}'))
ax.set_yticks(y, [f'{b} days' for b in WAIT[1:]]); ax.invert_yaxis(); ax.grid(axis='y', visible=False)
ax.set_xlabel('Odds ratio vs. no waiting (log scale, 95% CI)'); ax.legend(loc='lower right')
suptitle(fig, '3.3  Once booking profile is controlled for, waiting-list effects shrink and point in mixed directions',
         'Logistic regression: odds of cancellation by waiting-list band, before vs after controlling for booking profile')
plt.show()
pd.concat(res, axis=1).round(3)
```

Waiting list has no consistent effect on cancellation

### RO5 — High-volume, high-risk booking profiles

If the hotel only focus on a few types of customer, would cancellation rate
decrease?

NOTE:
- Profile is a combination of hotel + market segment + deposit type and lead-time band e.g., “City · Online TA · No Deposit · booked 6–12 months ahead”. The reason why i didnt include distribution channel as part of the profile its because it didnt move the needle for the share of all cancellations for the top 10 profile
- Priority profiles = the top 10 by **excess cancellations** among profiles with at least 1,000 bookings. Excess cancellations = bookings × (profile rate − 37% average): how many more cancellations a profile produces than an average group of the same size.

```python
pcols = ['hotel', 'market_segment', 'deposit_type', 'lead_band']
def build_profiles(data, cols, base):
    p = data.groupby(cols, observed=True)['is_canceled'].agg(bookings='size', cancellations='sum').reset_index()
    p['cancel_rate'] = p['cancellations'] / p['bookings']
    p['excess'] = p['bookings'] * (p['cancel_rate'] - base)
    p['profile'] = p[cols].astype(str).agg(' · '.join, axis=1).str.replace(' Hotel', '', regex=False) + 'd'
    return p.sort_values('excess', ascending=False).reset_index(drop=True)

prof_all = build_profiles(seg, pcols, BASE)
prof = prof_all[prof_all['bookings'] >= 300].copy()                    # shown on the map
top = prof_all[prof_all['bookings'] >= 1000].head(10).copy(); top['rank'] = range(1, len(top) + 1)
prof['priority'] = prof.index.isin(top.index)

# 5.1 Volume vs rate map
fig, ax = plt.subplots(figsize=(11, 6))
o = prof[~prof['priority']]
ax.scatter(o['bookings'], o['cancel_rate'], s=45, color=CONTEXT, edgecolor=SURFACE, linewidth=1.5, label='Other profiles (≥300 bookings)')
ax.scatter(top['bookings'], top['cancel_rate'], s=75, color=BLUE, edgecolor=SURFACE, linewidth=2, label='Top 10 by excess cancellations (≥1,000 bookings)')
for _, r in top.iterrows():
    ax.annotate(str(r['rank']), (r['bookings'], r['cancel_rate']), xytext=(6, 4), textcoords='offset points',
                fontsize=9, fontweight='bold')
ax.axhline(BASE, color=INK2, lw=1, ls=(0, (4, 3))); ax.axvline(1000, color=INK2, lw=1, ls=(0, (4, 3)))
ax.set_xscale('log')
ax.text(0.01, BASE + 0.015, f'average {BASE:.0%}', fontsize=8.5, color=INK2, transform=ax.get_yaxis_transform())
ax.text(1000 * 1.05, 0.02, '1,000 bookings', fontsize=8.5, color=INK2, transform=ax.get_xaxis_transform()); ax.xaxis.set_major_formatter(THOU); ax.yaxis.set_major_formatter(PCT); ax.set_ylim(0, 1.05)
ax.set_xlabel('Bookings in profile (log scale)'); ax.set_ylabel('Cancellation rate'); ax.legend(loc='lower left', bbox_to_anchor=(0, 1.0), ncol=2)
suptitle(fig, f'5.1  Ten profiles produce {top["cancellations"].sum() / TOTAL_CANCELS:.0%} of all cancellations',
         'Each dot = hotel × segment × deposit × lead-time profile; numbers match ranks in 5.2', gap=1.3)
plt.show()
```

```python
# 5.2 Ranked priority profiles by excess cancellations
t = top.iloc[::-1]
cols = [ORANGE if d == 'Non Refund' else BLUE for d in t['deposit_type']]
fig, ax = plt.subplots(figsize=(11, 0.48 * len(t) + 2.0))
ax.barh(range(len(t)), t['excess'], color=cols, height=0.62)
ax.set_yticks(range(len(t)), [f"{r}. {p}" for r, p in zip(t['rank'], t['profile'])])
for yi, (_, r) in enumerate(t.iterrows()):
    ax.text(r['excess'], yi, f"  {r['excess']:,.0f}   (n={r['bookings']:,}, {r['cancel_rate']:.0%})", va='center', fontsize=8.5, color=INK2)
ax.set_xlim(0, t['excess'].max() * 1.45); ax.xaxis.set_major_formatter(THOU); ax.grid(axis='y', visible=False)
ax.set_xlabel('Excess cancellations vs. overall rate')
from matplotlib.patches import Patch
ax.legend(handles=[Patch(color=ORANGE, label='Block release (Non Refund)'), Patch(color=BLUE, label='Guest booking (No Deposit)')],
          loc='lower right')
suptitle(fig, '5.2  Six block-release profiles, then long-lead Online TA bookings',
         f'Top 10 profiles with ≥1,000 bookings, ranked by cancellations above the {BASE:.0%} average')
fig.subplots_adjust(left=0.38); plt.show()
top[['rank', 'profile', 'bookings', 'cancel_rate', 'excess']].round(3)
```

> The orange bars in 5.2 are **block releases**: non-refundable Group and Offline TA/TO room blocks at the City Hotel, released almost in full. Their fix is contractual (release dates, penalties for unsold rooms), not guest-level. 5.3 repeats the ranking **excluding block releases**, where guest-level actions (deposits, reconfirmation, overbooking) apply.

```python
# 5.3 Priority profiles excluding block releases (excess measured against the lower 28% average)
prof_ex = build_profiles(seg[seg['block_release'] == 0], pcols, BASE_EX)
nd = prof_ex[(prof_ex['bookings'] >= 1000) & (prof_ex['excess'] > 0)].head(10).iloc[::-1]
lab = nd['profile'].str.replace(' · No Deposit', '', regex=False)
fig, ax = plt.subplots(figsize=(11, 0.48 * len(nd) + 1.8))
ax.barh(range(len(nd)), nd['excess'], color=BLUE, height=0.62)
ax.set_yticks(range(len(nd)), lab)
for yi, (_, r) in enumerate(nd.iterrows()):
    ax.text(r['excess'], yi, f"  {r['excess']:,.0f}   (n={r['bookings']:,}, {r['cancel_rate']:.0%})", va='center', fontsize=8.5, color=INK2)
ax.set_xlim(0, nd['excess'].max() * 1.45); ax.xaxis.set_major_formatter(THOU); ax.grid(axis='y', visible=False)
ax.set_xlabel(f'Excess cancellations vs. the {BASE_EX:.0%} average excluding block releases')
suptitle(fig, '5.3  Excluding block releases, long-lead Online TA reservations are the main target',
         f'Profiles (hotel · segment · lead time) with ≥1,000 bookings and a rate above {BASE_EX:.0%}, block releases excluded')
fig.subplots_adjust(left=0.3); plt.show()
nd.iloc[::-1][['profile', 'bookings', 'cancel_rate', 'excess']].round(3)
```

```python
# 5.4 Robustness checks — (a) other thresholds, (b) adding distribution channel to the profile
# (a) How many profiles would qualify, and what share of all cancellations they cover, under other cut-offs
rows = []
for r in [0.4, 0.5, 0.6]:
    row = {'Cancellation rate ≥': f'{r:.0%}'}
    for v in [500, 1000, 2000]:
        q = prof_all[(prof_all['cancel_rate'] >= r) & (prof_all['bookings'] >= v)]
        row[f'≥{v:,} bookings'] = f"{len(q)} profile{'s' if len(q) != 1 else ''} · {q['cancellations'].sum() / TOTAL_CANCELS:.0%} of cancellations"
    rows.append(row)
print('(a) Threshold sensitivity — profiles that would qualify under a fixed rate + volume cut-off')
print(f"    Current rule (top 10 by excess, ≥1,000 bookings): 10 profiles · {top['cancellations'].sum() / TOTAL_CANCELS:.0%} of cancellations")
display(pd.DataFrame(rows).set_index('Cancellation rate ≥'))

# (b) Same ranking with distribution channel added as a fifth part of the profile
pcols_ch = ['hotel', 'market_segment', 'distribution_channel', 'deposit_type', 'lead_band']
top_ch = build_profiles(seg, pcols_ch, BASE)
top_ch = top_ch[top_ch['bookings'] >= 1000].head(10).reset_index(drop=True)
cmp = pd.DataFrame({'Rank': range(1, 11),
                    'Profile (current)': top['profile'].values, 'Excess': top['excess'].round(0).astype(int).values,
                    'Profile with channel added': top_ch['profile'].values, 'Excess ': top_ch['excess'].round(0).astype(int).values})
print(f"(b) Adding distribution channel: {len(build_profiles(seg, pcols, BASE))} → {len(build_profiles(seg, pcols_ch, BASE))} profiles; "
      f"top 10 cover {top['cancellations'].sum() / TOTAL_CANCELS:.0%} vs {top_ch['cancellations'].sum() / TOTAL_CANCELS:.0%} of cancellations")
display(cmp.set_index('Rank'))
```

The 10 highest-impact booking profiles, each with at least 1,000 bookings, account for 47% of all cancellations. Six are non-refundable Groups or Offline TA/TO block-release bookings at the City Hotel, with cancellation rates close to 100%. The other four are no-deposit Online TA bookings made more than 90 days in advance, at both hotels. Excluding block releases, City Hotel Online TA bookings made 31 to 365 days in advance become the biggest concern. The ranking is unchanged when different rate and volume thresholds are used or when distribution channel is added to the profile.

# Part 2: Predictive Analysis

## Step 1. Import libraries and define target

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
import seaborn as sns

target = 'is_canceled'
features = ['deposit_type', 'lead_time', 'market_segment', 'total_of_special_requests', 'hotel']

X = pd.get_dummies(df[features], dtype=int)   # one-hot encode the text columns
y = df[target]
X.head()
```

## Step 2. Split into training and testing data

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=0, stratify=y)
print('Training rows:', len(X_train))
print('Testing rows :', len(X_test))
```

## Step 3. Train and evaluate the models

```python
models = {
    'Logistic Regression': LogisticRegression(max_iter=1000),
    'Decision Tree': DecisionTreeClassifier(max_depth=6, random_state=0),
    'Random Forest': RandomForestClassifier(n_estimators=100, max_depth=10, random_state=0),
}

accuracy = {}
for name, model in models.items():
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    accuracy[name] = accuracy_score(y_test, y_pred)

    print(f'===== {name} =====')
    print(classification_report(y_test, y_pred, target_names=['Not cancelled', 'Cancelled']))
    sns.heatmap(confusion_matrix(y_test, y_pred), annot=True, fmt='d', cmap='Blues')
    plt.title(f'{name}: confusion matrix'); plt.xlabel('Predicted'); plt.ylabel('Actual')
    plt.show()
```

## Step 4. Compare the models

```python
compare = pd.DataFrame({'Model': accuracy.keys(), 'Accuracy': accuracy.values()})
compare = compare.sort_values('Accuracy', ascending=False)

baseline = 1 - y_test.mean()   # accuracy if we always guess "not cancelled"
print(f'Baseline accuracy (always guess "not cancelled"): {baseline:.1%}')
compare.style.format({'Accuracy': '{:.1%}'})
```

## Step 5. Feature importance using random forest



```python
rf = models['Random Forest']
importance = pd.Series(rf.feature_importances_, index=X.columns).sort_values()

importance.plot(kind='barh', figsize=(8, 5), color='#2a78d6')
plt.title('Feature importance (Random Forest)')
plt.xlabel('Importance')
plt.show()
```

## Step 6. Predict a new booking

```python
new_booking = pd.DataFrame([{
    'deposit_type': 'No Deposit',
    'lead_time': 200,
    'market_segment': 'Online TA',
    'total_of_special_requests': 0,
    'hotel': 'City Hotel',
}])
new_X = pd.get_dummies(new_booking, dtype=int).reindex(columns=X.columns, fill_value=0)

prob = rf.predict_proba(new_X)[0, 1]
print(f'Chance this booking will be cancelled: {prob:.0%}')
```

- Random Forest is the most accurate at 79.6% followed by Decision Tree (79.0%) and Logistic Regression (78.0%).
- Deposit type and lead time are the most important features followed by special requests and the Online TA segment e.g., a City Hotel, Online TA, No Deposit booking made 200 days ahead with no special requests has a 68% predicted chance of cancelling.



Limitations: only two hotels; five features; the model shows association, not cause.

```python
# Step 6. Compare different feature sets
feature_sets = {
    'Set A: booking terms (5)': ['deposit_type', 'lead_time', 'market_segment',
                                 'total_of_special_requests', 'hotel'],
    'Set B: stay & channel (6)': ['lead_time', 'total_nights', 'is_repeated_guest',
                                  'market_segment', 'distribution_channel', 'hotel'],
    'Set C: guest history (5)': ['is_repeated_guest', 'previous_cancellations',
                                 'previous_bookings_not_canceled', 'customer_type', 'adults'],
    'Set D: A + B + C combined': ['deposit_type', 'lead_time', 'market_segment',
                                  'total_of_special_requests', 'hotel', 'total_nights',
                                  'is_repeated_guest', 'distribution_channel',
                                  'previous_cancellations', 'previous_bookings_not_canceled',
                                  'customer_type', 'adults'],
}

results = []
for name, cols in feature_sets.items():
    X_set = pd.get_dummies(df[cols], dtype=int)
    Xtr, Xte, ytr, yte = train_test_split(X_set, y, test_size=0.3, random_state=0, stratify=y)
    model = RandomForestClassifier(n_estimators=100, max_depth=10, random_state=0)
    model.fit(Xtr, ytr)
    results.append({'Feature set': name, 'No. of features': len(cols),
                    'Accuracy': accuracy_score(yte, model.predict(Xte))})

fs_compare = pd.DataFrame(results)
fs_compare.plot(x='Feature set', y='Accuracy', kind='barh', legend=False, figsize=(8, 4), color='#2a78d6')
plt.axvline(1 - y.mean(), color='grey', ls='--', label='Baseline')
plt.xlim(0.5, 0.85); plt.xlabel('Accuracy'); plt.title('Accuracy by feature set (Random Forest)')
plt.legend(); plt.show()
fs_compare.style.format({'Accuracy': '{:.1%}'})
```
