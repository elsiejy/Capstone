## 3. Data Cleansing & Preparation

Let's clean the dataset. We'll identify missing values, handle empty rows, correct data types, and prepare the dataset for analysis.

Data cleansing on booking dataset:
1. Derive check-in date
2. Update data type for reservation_status_date
3. Update data type for is_repeated_guest and is_canceled
4. Update data type for children, agent and company
5. Handle NULL values for children, country, agent, company

```python
# 1. Derive check-in date
# combine arrival_date_year, arrival_date_month and arrival_date_day_of_month into strings
booking_date_strings = (
    bookings['arrival_date_year'].astype(str) + '-' +
    bookings['arrival_date_month'].astype(str) + '-' +
    bookings['arrival_date_day_of_month'].astype(str)
)

# convert the strings into a proper pandas datetime object
bookings['checkin_date'] = pd.to_datetime(booking_date_strings)

# verify that first 5 rows of checkin_date
print(bookings['checkin_date'].head())
```

```python
# 2. Update data type for reservation_status_date
# convert reservation_status_date to datetime format
bookings['reservation_status_date'] = pd.to_datetime(bookings['reservation_status_date'])

# view the first 5 rows of reservation_status_date
print(bookings['reservation_status_date'].head())
```

```python
# 3. Update data type for is_repeated_guest and is_canceled
# change datatype from int to boolean
bookings[['is_repeated_guest', 'is_canceled']] = bookings[['is_repeated_guest', 'is_canceled']].astype(bool)

# verify count of is_repeated_guest and is_canceled
print(bookings['is_canceled'].value_counts())
print(bookings['is_repeated_guest'].value_counts())

# confirm values are only false, true
print(bookings['is_canceled'].unique())
print(bookings['is_repeated_guest'].unique())
```

```python
# 4. Update data type for children, agent and company
# change datatype from float to int
bookings[['children', 'agent', 'company']] = bookings[['children', 'agent', 'company']].astype('Int64')
```

```python
# 5. Handle NULL values for children, country, agent, company
# handle NULL value for children by replacing NULL with 0
bookings['children'] = bookings['children'].fillna(0)

# handle NULL value for country by replacing NULL with UNKNOWN
bookings['country'] = bookings['country'].fillna('Unknown')

# handle NULL value for agent by replacing NULL with 0
bookings['agent'] = bookings['agent'].fillna(0)

# handle NULL value for company by replacing NULL with 0
bookings['company'] = bookings['company'].fillna(0)
```

### 0.1 Data check

`lead_time` counts **backwards** from arrival, so `booking_date = checkin_date − lead_time`. `reservation_status_date` is the date the *last* status was set, so its meaning depends on the status:

* **Check-Out** → the checkout date (≈ check-in + total nights)
* **Canceled** → the cancellation date
* **No-Show** → the arrival date

Example — Booking 9: booked 7 Apr 2015 (1 Jul − 85 days), cancelled 6 May 2015, due to arrive 1 Jul 2015. It was cancelled 29 days after booking and **56 days before arrival**. The check below confirms the rule holds across the whole file.

```python
gap = (df['reservation_status_date'] - df['checkin_date']).dt.days
co = df['reservation_status'].eq('Check-Out')
ca = df['reservation_status'].eq('Canceled')
ns = df['reservation_status'].eq('No-Show')

checks = pd.DataFrame({
    'rows': [co.sum(), ca.sum(), ns.sum()],
    'rule': ['status date = check-in + nights', 'cancelled on/before arrival and on/after booking',
             'status date = check-in'],
    'rows_meeting_rule': [(gap[co] == df.loc[co, 'total_nights']).sum(),
                          ((gap[ca] <= 0) & (df.loc[ca, 'reservation_status_date'] >= df.loc[ca, 'booking_date'])).sum(),
                          (gap[ns] == 0).sum()],
}, index=['Check-Out', 'Canceled', 'No-Show'])
checks['% meeting rule'] = (checks['rows_meeting_rule'] / checks['rows']).map('{:.2%}'.format)
display(checks)
print(df.loc[df['Booking_ID'] == 9, ['booking_date', 'reservation_status_date', 'checkin_date',
                                    'reservation_status', 'days_before_cancel']].to_string(index=False))
```

The dates are consistent: 99.96% of check-outs leave on check-in + nights (the 27 exceptions are same-day departures), every cancellation is dated between booking and arrival, and no-shows are recorded on the arrival date. `reservation_status_date` can therefore be used as the **cancellation date**

```python
def summarise(d, by):
    # Volume, cancellation and revenue metrics for any grouping.
    t = d.groupby(by, observed=True).agg(
        bookings=('is_canceled', 'size'), cancellations=('is_canceled', 'sum'),
        booked_rev=('room_revenue', 'sum'), rev_at_risk=('rev_at_risk', 'sum'),
        late_rev_at_risk=('late_rev_at_risk', 'sum'), median_adr=('adr', 'median'),
        avg_nights=('total_nights', 'mean'))
    t['cancel_rate'] = t['cancellations'] / t['bookings']
    t['rev_loss_rate'] = t['rev_at_risk'] / t['booked_rev']
    t['share_bookings'] = t['bookings'] / len(d)
    t['share_cancellations'] = t['cancellations'] / d['is_canceled'].sum()
    t['share_rev_at_risk'] = t['rev_at_risk'] / d['rev_at_risk'].sum()
    t['late_share_of_risk'] = t['late_rev_at_risk'] / t['rev_at_risk']
    return t

def show(t, cols=None, sort=None, n=None):
    # Display a summary table with readable %, $ and count formats.
    t = t.copy()
    if sort: t = t.sort_values(sort, ascending=False)
    if n: t = t.head(n)
    cols = cols or list(t.columns)
    out = pd.DataFrame(index=t.index)
    for c in cols:
        v = t[c]
        if c.endswith('_rate') or c.startswith('share_') or c.endswith('_share') or c.endswith('_of_risk') or c.startswith('pct'):
            out[c] = v.map(lambda x: f'{x:.1%}' if pd.notna(x) else '')
        elif (c.startswith('rev') or c.endswith('_rev') or 'rev_at_risk' in c) and not c.startswith('rank'):
            out[c] = v.map(lambda x: f'${x/1e6:,.2f}M' if abs(x) >= 1e5 else f'${x:,.0f}')
        elif c == 'median_adr':
            out[c] = v.map(lambda x: f'${x:,.0f}')
        elif not pd.api.types.is_numeric_dtype(v):
            out[c] = v
        elif pd.api.types.is_float_dtype(v):
            out[c] = v.map(lambda x: f'{x:,.1f}')
        else:
            out[c] = v.map(lambda x: f'{x:,}')
    display(out)

def three_panel(t, title, figsize=(12, 3.2), order=None, overall=None):
    # Bookings | cancellation rate | revenue at risk, as small multiples on one category axis.
    t = t.loc[order] if order is not None else t.sort_values('rev_at_risk')
    labels = [str(i) for i in t.index]
    fig, axes = plt.subplots(1, 3, figsize=figsize, sharey=True)
    specs = [('bookings', NEUTRAL, 'Bookings', lambda x: f'{x:,.0f}'),
             ('cancel_rate', CANC, 'Cancellation rate', lambda x: f'{x:.0%}'),
             ('rev_at_risk', CANC, 'Revenue at risk', lambda x: f'${x/1e6:,.1f}M')]
    for ax, (col, colr, name, f) in zip(axes, specs):
        ax.barh(labels, t[col], color=colr, height=0.62)
        for y, v in enumerate(t[col]):
            ax.text(v, y, ' ' + f(v), va='center', fontsize=8.5, color=INK2)
        ax.set_title(name, fontsize=10)
        ax.grid(axis='y', visible=False)
        ax.margins(x=0.22)
        ax.xaxis.set_major_locator(mtick.MaxNLocator(4))
        if col == 'cancel_rate':
            ax.xaxis.set_major_formatter(PCT)
            if overall is not None:
                ax.axvline(overall, color=INK2, ls='--', lw=0.8)
        if col == 'rev_at_risk': ax.xaxis.set_major_formatter(EUR_M)
    fig.suptitle(title, x=0.01, ha='left', fontweight='bold', fontsize=12)
    fig.tight_layout()
    plt.show()

def cramers_v(x, y):
    ct = pd.crosstab(x, y)
    chi2 = chi2_contingency(ct, correction=False)[0]
    n = ct.values.sum(); r, k = ct.shape
    return np.sqrt(chi2 / (n * (min(r, k) - 1)))

OVERALL_RATE = df['is_canceled'].mean()
OVERALL_LOSS = df['rev_at_risk'].sum() / df['room_revenue'].sum()
```

#### Chart 2.3 — Ranking the strength of association
Cramér's V (0 = no association, 1 = perfect) for each characteristic against cancellation, computed on **all bookings** and on bookings **excluding block releases** (which otherwise dominate deposit type and segment). The share of revenue at risk held by the riskiest category of each factor is shown alongside.

```python
factors = {
    'Deposit type': 'deposit_type', 'Lead time': 'lead_bucket', 'Previous cancellations': 'prev_cancel_bucket',
    'Market segment': 'market_segment', 'Special requests*': 'requests_bucket', 'Distribution channel': 'distribution_channel',
    'Customer type': 'customer_type', 'Booking changes*': 'changes_bucket', 'Hotel': 'hotel', 'Party composition': 'party_type',
    'Length of stay': 'stay_bucket', 'Repeat guest': 'is_repeated_guest', 'Meal plan': 'meal', 'Waiting list': 'waitlist_bucket',
}
nb = df[~df['block_release']]
rank = pd.DataFrame({
    'cramers_v_all': {k: cramers_v(df[v], df['is_canceled']) for k, v in factors.items()},
    'cramers_v_excl_block': {k: cramers_v(nb[v], nb['is_canceled']) for k, v in factors.items()},
}).sort_values('cramers_v_all', ascending=False)
top_cat = {}
for k, v in factors.items():
    t = summarise(nb, v)
    t = t[t['bookings'] >= 500]
    worst = t['cancel_rate'].idxmax()
    top_cat[k] = f"{worst} ({t.loc[worst, 'cancel_rate']:.0%}, {t.loc[worst, 'share_rev_at_risk']:.0%} of risk)"
rank['riskiest category, excl. block releases (rate, share of revenue at risk)'] = pd.Series(top_cat)
display(rank.round(3))

fig, ax = plt.subplots(figsize=(10, 5))
r = rank.sort_values('cramers_v_all')
yy = np.arange(len(r))
ax.barh(yy + 0.2, r['cramers_v_all'], 0.38, color=CANC, label='All bookings')
ax.barh(yy - 0.2, r['cramers_v_excl_block'], 0.38, color=NEUTRAL, label='Excluding block releases')
ax.set_yticks(yy); ax.set_yticklabels(r.index); ax.grid(axis='y', visible=False)
ax.set_xlabel("Cramér's V with cancellation"); ax.legend(loc='lower right')
ax.set_title('Chart 2.3  Without block releases, lead time, segment and previous cancellations lead\n(* snapshot variable — partly a timing effect)', loc='left')
plt.tight_layout(); plt.show()
```

**Finding:** On all bookings, **deposit type** is the strongest single signal (V = 0.48), followed by lead time, previous cancellations, market segment and special requests. Excluding block releases, deposit type drops to almost nothing (almost every remaining booking is No Deposit), and the leaders become **lead time (0.23), market segment (0.21) and previous cancellations (0.20)**. The snapshot variables (special requests, booking changes) fall noticeably once block releases are removed. Waiting list and repeat-guest status are weak.

```python
m_df = df[df['market_segment'] != 'Undefined'].copy()
m_df['waitlisted'] = (m_df['days_in_waiting_list'] > 0).astype(int)
m_df['wl_days_30'] = m_df['days_in_waiting_list'] / 30
m_df['prev_cancel'] = (m_df['previous_cancellations'] > 0).astype(int)
m_df['lead_bucket'] = m_df['lead_bucket'].astype(str)
controls = 'C(hotel) + C(market_segment) + C(lead_bucket) + C(customer_type) + prev_cancel + is_repeated_guest'

def odds(formula, data, term):
    res = smf.logit(formula, data=data).fit(disp=0, maxiter=200)
    ci = np.exp(res.conf_int().loc[term])
    return pd.Series({'odds_ratio': np.exp(res.params[term]), 'ci_low': ci[0], 'ci_high': ci[1], 'p_value': res.pvalues[term]})

nd = m_df[m_df['deposit_type'] == 'No Deposit']
or_tbl = pd.DataFrame({
    'Raw (all bookings)': odds('is_canceled ~ waitlisted', m_df, 'waitlisted'),
    'Adjusted (all, + deposit type)': odds(f'is_canceled ~ waitlisted + C(deposit_type) + {controls}', m_df, 'waitlisted'),
    'Adjusted (No Deposit only)': odds(f'is_canceled ~ waitlisted + {controls}', nd, 'waitlisted'),
    'Adjusted, per 30 days waited (No Deposit)': odds(f'is_canceled ~ wl_days_30 + {controls}', nd, 'wl_days_30'),
}).T
display(or_tbl.round(3))

fig, ax = plt.subplots(figsize=(8, 2.8))
yy = np.arange(len(or_tbl))[::-1]
ax.errorbar(or_tbl['odds_ratio'], yy, xerr=[or_tbl['odds_ratio'] - or_tbl['ci_low'], or_tbl['ci_high'] - or_tbl['odds_ratio']],
            fmt='o', color=CANC, ecolor=INK2, capsize=3, ms=7)
ax.axvline(1, color=INK2, ls='--', lw=0.8); ax.set_xscale('log')
ax.set_xticks([0.25, 0.5, 1, 2, 4]); ax.set_xticklabels(['0.25', '0.5', '1', '2', '4']); ax.xaxis.set_minor_formatter(mtick.NullFormatter())
ax.set_yticks(yy); ax.set_yticklabels(or_tbl.index); ax.grid(axis='y', visible=False)
ax.set_xlabel('Odds ratio for cancellation (95% CI, log scale)')
ax.set_title('Chart 3.3  The raw waiting-list effect disappears once deposit type is controlled for', loc='left')
plt.tight_layout(); plt.show()
```

**Finding (Charts 3.1–3.3):** Waiting-list bookings look risky in raw terms (53–76% vs 36%), but 34–71% of them are Non Refund. Like for like they cancel **less**: City No Deposit 31% → 23%, Resort No Deposit 25% → 5%. In logistic regression, the raw odds ratio of 3.1 falls to **1.00** (95% CI 0.88–1.13) once deposit type and other characteristics are controlled for, and to **0.95** within No Deposit bookings. Each additional 30 days waited is associated with a small increase (OR 1.05). **Waiting-list days are not an independent risk driver**, and waiting-list bookings hold only 3.6% of revenue at risk.

## RO5 — High-risk booking profiles: by count and by revenue
A profile is **hotel × market segment × deposit type × lead-time band**, kept if it has at least 1,000 bookings.

* **Excess cancellations** = cancellations − (bookings × overall cancellation rate)
* **Excess revenue at risk** = revenue at risk − (booked revenue × overall revenue loss rate)

Both measure how much worse a profile is than average, scaled by its size. Ranking by the second answers: *where would fixing the problem recover the most money?*

```python
PROFILE = ['hotel', 'market_segment', 'deposit_type', 'lead_bucket']
H_SHORT = {'City Hotel': 'City', 'Resort Hotel': 'Resort'}

def profile_rank(d, keys=PROFILE, min_n=1000):
    rate = d['is_canceled'].mean(); loss = d['rev_at_risk'].sum() / d['room_revenue'].sum()
    t = summarise(d, keys)
    t = t[t['bookings'] >= min_n].copy()
    t['excess_canc'] = t['cancellations'] - t['bookings'] * rate
    t['excess_rev_at_risk'] = t['rev_at_risk'] - t['booked_rev'] * loss
    t['rank_by_count'] = t['excess_canc'].rank(ascending=False).astype(int)
    t['rank_by_revenue'] = t['excess_rev_at_risk'].rank(ascending=False).astype(int)
    t.index = [' · '.join([H_SHORT.get(i[0], i[0])] + [str(x) for x in i[1:]]) + ('d' if 'lead_bucket' in keys else '') for i in t.index]
    return t

prof = profile_rank(df)
print(f'{len(prof)} profiles with ≥1,000 bookings, covering {prof["bookings"].sum()/len(df):.0%} of bookings')
top_c = prof.nsmallest(10, 'rank_by_count'); top_r = prof.nsmallest(10, 'rank_by_revenue')
print(f"Top 10 by excess cancellations: {top_c['share_cancellations'].sum():.0%} of cancellations, {top_c['share_rev_at_risk'].sum():.0%} of revenue at risk")
print(f"Top 10 by excess revenue:       {top_r['share_cancellations'].sum():.0%} of cancellations, {top_r['share_rev_at_risk'].sum():.0%} of revenue at risk")
print(f"Profiles in both top-10 lists: {len(set(top_c.index) & set(top_r.index))}")
```

```python
daily = c.groupby('reservation_status_date').agg(cancellations=('is_canceled', 'size'), rev=('room_revenue', 'sum'),
                                                 pct_non_refund=('block_release', 'mean'),
                                                 main_segment=('market_segment', lambda s: s.value_counts().index[0]))
top_days = daily.nlargest(10, 'cancellations')
daily_full = daily['cancellations'].reindex(pd.date_range(daily.index.min(), daily.index.max()), fill_value=0)
print(f"Median cancellations recorded per day: {daily['cancellations'].median():.0f}")
print(f"Top 10 days hold {top_days['cancellations'].sum()/len(c):.1%} of all cancellations and {top_days['rev'].sum()/c['room_revenue'].sum():.1%} of revenue at risk")
show(top_days)

fig, ax = plt.subplots(figsize=(12, 3))
ax.plot(daily_full.index, daily_full.values, color=NEUTRAL, lw=0.7)
ax.scatter(top_days.index, top_days['cancellations'], color=CANC, s=30, zorder=3, label='Top 10 days')
ax.set_ylabel('Cancellations recorded'); ax.legend(loc='upper left')
ax.set_title('Chart 6.4  Cancellations recorded per day — spikes are bulk releases of group allocations', loc='left')
plt.tight_layout(); plt.show()
```

**Finding:** On a typical day 36 cancellations are recorded, but on a few days hundreds are. The top 10 days hold **13.7%** of all cancellations (7.4% of revenue at risk), and on most of them 70–99% are Non Refund Group or Offline TA/TO bookings. This supports reading Non Refund cancellations as **bulk allocation releases** by agents rather than individual guest decisions.

# Part 2: Predictive Analysis

## Step 1. Import libraries and define target

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
import seaborn as sns

target = 'is_canceled'
features = ['deposit_type', 'lead_time', 'market_segment', 'total_of_special_requests', 'hotel']

X = pd.get_dummies(df[features], dtype=int)   # one-hot encode the text columns
y = df[target]
X.head()
```

## Step 2. Split into training and testing data

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=0, stratify=y)
print('Training rows:', len(X_train))
print('Testing rows :', len(X_test))
```

## Step 3. Train and evaluate the models

```python
models = {
    'Logistic Regression': LogisticRegression(max_iter=1000),
    'Decision Tree': DecisionTreeClassifier(max_depth=6, random_state=0),
    'Random Forest': RandomForestClassifier(n_estimators=100, max_depth=10, random_state=0),
}

accuracy = {}
for name, model in models.items():
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    accuracy[name] = accuracy_score(y_test, y_pred)

    print(f'===== {name} =====')
    print(classification_report(y_test, y_pred, target_names=['Not cancelled', 'Cancelled']))
    sns.heatmap(confusion_matrix(y_test, y_pred), annot=True, fmt='d', cmap='Blues')
    plt.title(f'{name}: confusion matrix'); plt.xlabel('Predicted'); plt.ylabel('Actual')
    plt.show()
```

## Step 4. Compare the models

```python
compare = pd.DataFrame({'Model': accuracy.keys(), 'Accuracy': accuracy.values()})
compare = compare.sort_values('Accuracy', ascending=False)

baseline = 1 - y_test.mean()   # accuracy if we always guess "not cancelled"
print(f'Baseline accuracy (always guess "not cancelled"): {baseline:.1%}')
compare.style.format({'Accuracy': '{:.1%}'})
```

## Step 5. Feature importance using random forest



```python
rf = models['Random Forest']
importance = pd.Series(rf.feature_importances_, index=X.columns).sort_values()

importance.plot(kind='barh', figsize=(8, 5), color='#2a78d6')
plt.title('Feature importance (Random Forest)')
plt.xlabel('Importance')
plt.show()
```

## Step 6. Predict a new booking

```python
new_booking = pd.DataFrame([{
    'deposit_type': 'No Deposit',
    'lead_time': 200,
    'market_segment': 'Online TA',
    'total_of_special_requests': 0,
    'hotel': 'City Hotel',
}])
new_X = pd.get_dummies(new_booking, dtype=int).reindex(columns=X.columns, fill_value=0)

prob = rf.predict_proba(new_X)[0, 1]
print(f'Chance this booking will be cancelled: {prob:.0%}')
```

- Random Forest is the most accurate at 79.6% followed by Decision Tree (79.0%) and Logistic Regression (78.0%).
- Deposit type and lead time are the most important features followed by special requests and the Online TA segment e.g., a City Hotel, Online TA, No Deposit booking made 200 days ahead with no special requests has a 68% predicted chance of cancelling.



Limitations: only two hotels; five features; the model shows association, not cause.

```python
# Step 6. Compare different feature sets
feature_sets = {
    'Set A: booking terms (5)': ['deposit_type', 'lead_time', 'market_segment',
                                 'total_of_special_requests', 'hotel'],
    'Set B: stay & channel (6)': ['lead_time', 'total_nights', 'is_repeated_guest',
                                  'market_segment', 'distribution_channel', 'hotel'],
    'Set C: guest history (5)': ['is_repeated_guest', 'previous_cancellations',
                                 'previous_bookings_not_canceled', 'customer_type', 'adults'],
    'Set D: A + B + C combined': ['deposit_type', 'lead_time', 'market_segment',
                                  'total_of_special_requests', 'hotel', 'total_nights',
                                  'is_repeated_guest', 'distribution_channel',
                                  'previous_cancellations', 'previous_bookings_not_canceled',
                                  'customer_type', 'adults'],
}

results = []
for name, cols in feature_sets.items():
    X_set = pd.get_dummies(df[cols], dtype=int)
    Xtr, Xte, ytr, yte = train_test_split(X_set, y, test_size=0.3, random_state=0, stratify=y)
    model = RandomForestClassifier(n_estimators=100, max_depth=10, random_state=0)
    model.fit(Xtr, ytr)
    results.append({'Feature set': name, 'No. of features': len(cols),
                    'Accuracy': accuracy_score(yte, model.predict(Xte))})

fs_compare = pd.DataFrame(results)
fs_compare.plot(x='Feature set', y='Accuracy', kind='barh', legend=False, figsize=(8, 4), color='#2a78d6')
plt.axvline(1 - y.mean(), color='grey', ls='--', label='Baseline')
plt.xlim(0.5, 0.85); plt.xlabel('Accuracy'); plt.title('Accuracy by feature set (Random Forest)')
plt.legend(); plt.show()
fs_compare.style.format({'Accuracy': '{:.1%}'})
```
