## 3. Data Cleansing & Preparation

Let's clean the dataset. We'll identify missing values, handle empty rows, correct data types, and prepare the dataset for analysis.

Data cleansing on booking dataset:
1. Derive check-in date
2. Update data type for reservation_status_date
3. Update data type for is_repeated_guest and is_canceled
4. Update data type for children, agent and company
5. Handle NULL values for children, country, agent, company

```python
# 1. Derive check-in date
# combine arrival_date_year, arrival_date_month and arrival_date_day_of_month into strings
booking_date_strings = (
    bookings['arrival_date_year'].astype(str) + '-' +
    bookings['arrival_date_month'].astype(str) + '-' +
    bookings['arrival_date_day_of_month'].astype(str)
)

# convert booking_date_strings to datetime then format as YYYY-MM-DD text
bookings['checkin_date'] = pd.to_datetime(booking_date_strings).dt.strftime('%Y-%m-%d')

# convert that string into a proper pandas datetime object
bookings['checkin_date'] = pd.to_datetime(booking_date_strings)

# verify that first 5 rows of checkin_date
print(bookings['checkin_date'].head())


```

```python
# 2. Update data type for reservation_status_date
# convert reservation_status_date to datetime format
bookings['reservation_status_date'] = pd.to_datetime(bookings['reservation_status_date'])

# view the first 5 rows of reservation_status_date
print(bookings['reservation_status_date'].head())
```

```python
# 3. Update data type for is_repeated_guest and is_canceled
# change datatype from int to boolean
bookings[['is_repeated_guest', 'is_canceled']] = bookings[['is_repeated_guest', 'is_canceled']].astype(bool)

# verify count of is_repeated_guest and is_canceled
print(bookings['is_canceled'].value_counts())
print(bookings['is_repeated_guest'].value_counts())

# confirm values are only false, true
print(bookings['is_canceled'].unique())
print(bookings['is_repeated_guest'].unique())
```

```python
# 4. Update data type for children, agent and company
# change datatype from float to int
bookings[['children', 'agent', 'company']] = bookings[['children', 'agent', 'company']].astype('Int64')
```

```python
# 5. Handle NULL values for children, country, agent, company
# handle NULL value for children by replacing NULL with 0
bookings['children'] = bookings['children'].fillna(0)

# handle NULL value for country by replacing NULL with UNKNOWN
bookings['country'] = bookings['country'].fillna('Unknown')

# handle NULL value for agent by replacing NULL with 0
bookings['agent'] = bookings['agent'].fillna(0)

# handle NULL value for company by replacing NULL with 0
bookings['company'] = bookings['company'].fillna(0)
```

## 4. Visualization for Research Objective

```python
import warnings; warnings.filterwarnings('ignore')
import matplotlib.pyplot as plt, matplotlib.ticker as mtick
from matplotlib.colors import LinearSegmentedColormap
from scipy.stats import chi2_contingency

# ---- Chart style: one validated palette, recessive grid, no dual axes ----
BLUE, ORANGE, AQUA = '#2a78d6', '#eb6834', '#1baf7a'          # categorical slots 1–3
INK, INK2, MUTED   = '#0b0b0b', '#52514e', '#8a8984'
CONTEXT = '#c9c7c0'                                            # neutral bars for volume / context
GRID, SURFACE = '#e6e5e1', '#fcfcfb'
HOTEL_COLORS = {'City Hotel': BLUE, 'Resort Hotel': ORANGE}
SEQ = LinearSegmentedColormap.from_list('seq_blue',
        ['#cde2fb', '#9ec5f4', '#6da7ec', '#3987e5', '#256abf', '#184f95', '#0d366b'])

plt.rcParams.update({
    'figure.facecolor': SURFACE, 'axes.facecolor': SURFACE, 'savefig.facecolor': SURFACE,
    'axes.edgecolor': GRID, 'axes.labelcolor': INK2, 'xtick.color': INK2, 'ytick.color': INK2,
    'text.color': INK, 'axes.titlecolor': INK, 'axes.titlesize': 11, 'axes.titleweight': 'bold',
    'axes.titlelocation': 'left', 'axes.spines.top': False, 'axes.spines.right': False,
    'axes.grid': True, 'grid.color': GRID, 'grid.linewidth': 0.8, 'axes.axisbelow': True,
    'font.size': 10, 'legend.frameon': False, 'figure.dpi': 110, 'lines.linewidth': 2,
    'xtick.major.size': 0, 'ytick.major.size': 0})
PCT = mtick.PercentFormatter(1.0, decimals=0)
THOU = mtick.FuncFormatter(lambda v, _: f'{v:,.0f}')

def suptitle(fig, text, sub=None, gap=1.0):
    # Title + subtitle anchored in inches from the top, so they never collide with the plot
    H = fig.get_figheight()
    fig.text(0.01, 1 - 0.10 / H, text, ha='left', va='top', fontsize=13, fontweight='bold')
    if sub: fig.text(0.01, 1 - 0.38 / H, sub, ha='left', va='top', fontsize=9.5, color=INK2)
    fig.subplots_adjust(top=1 - gap / H)
```

```python
# Source: the cleansed DataFrame `df` from the data-cleansing section of this notebook.
# The cleansing step already created every derived column used below (bands, flags, arrival fields),
# so this cell only checks they are present. Arrival-period fields come from checkin_date.
REQUIRED = ['is_canceled', 'checkin_date', 'hotel', 'market_segment', 'distribution_channel', 'customer_type',
            'deposit_type', 'meal', 'reservation_status', 'arrival_month', 'arrival_week', 'arrival_mname',
            'total_nights', 'kids', 'lead_band', 'stay_band', 'changes_band', 'requests_band', 'wait_band',
            'waited', 'party', 'prev_cancel_flag', 'prev_stay_flag', 'has_agent', 'has_company', 'country_prt']
missing = [c for c in REQUIRED if c not in df.columns]
assert not missing, f'These columns are missing from df — re-run the data-cleansing section: {missing}'

# Category orders, used to sort chart axes (must match the labels created during cleansing)
MONTHS = ['Jan','Feb','Mar','Apr','May','Jun','Jul','Aug','Sep','Oct','Nov','Dec']
LEAD  = ['0–7', '8–30', '31–90', '91–180', '181–365', '366+']
STAY  = ['0', '1', '2–3', '4–7', '8–14', '15+']
CHG   = ['0', '1', '2', '3+']
REQ   = ['0', '1', '2', '3+']
WAIT  = ['0', '1–7', '8–30', '31–90', '91+']
PARTY = ['Solo adult', 'Couple (2 adults)', '3+ adults', 'With children', 'No adults / unknown']

# 7 bookings carry an 'Undefined' segment/channel — excluded from segment-level views
seg = df[(df['market_segment'] != 'Undefined') & (df['distribution_channel'] != 'Undefined')].copy()

BASE = df['is_canceled'].mean()
print(f"Bookings: {len(df):,}   Cancelled + no-show: {df['is_canceled'].sum():,}   Baseline rate: {BASE:.2%}")
print(df['reservation_status'].value_counts().to_string())
```

```python
def rate_table(data, col, order=None):
    t = data.groupby(col, observed=True)['is_canceled'].agg(bookings='size', cancellations='sum')
    t['cancel_rate'] = t['cancellations'] / t['bookings']
    if order is not None:
        t = t.reindex([o for o in order if o in t.index])
    return t

def baseline_v(ax, label=True, y=None):
    ax.axvline(BASE, color=INK2, lw=1, ls=(0, (4, 3)), zorder=3)
    if label:
        ax.text(BASE, ax.get_ylim()[1] if y is None else y, f' overall {BASE:.1%}',
                color=INK2, fontsize=8.5, va='bottom', ha='left')

def baseline_h(ax, label=True):
    ax.axhline(BASE, color=INK2, lw=1, ls=(0, (4, 3)), zorder=3)
    if label:
        ax.text(ax.get_xlim()[1], BASE, f'overall {BASE:.1%} ', color=INK2, fontsize=8.5,
                va='bottom', ha='right')

def volume_rate_panel(data, col, title, sub=None, order=None):
    # Two aligned panels (one measure each, no dual axis): volume as context, rate as the story
    t = rate_table(data, col, order)
    if order is None:
        t = t.sort_values('cancel_rate')
    y = np.arange(len(t))
    fig, (a1, a2) = plt.subplots(1, 2, figsize=(11, 0.5 * len(t) + 1.9), sharey=True,
                                 gridspec_kw={'width_ratios': [1, 1.25], 'wspace': 0.12})
    a1.barh(y, t['bookings'], color=CONTEXT, height=0.62)
    a1.set_yticks(y, t.index); a1.set_title('Bookings'); a1.xaxis.set_major_formatter(THOU); a1.xaxis.set_major_locator(mtick.MaxNLocator(4))
    a1.set_xlim(0, t['bookings'].max() * 1.28); a1.grid(axis='y', visible=False)
    for yi, v in zip(y, t['bookings']):
        a1.text(v, yi, f' {v:,}', va='center', fontsize=8.5, color=INK2)
    a2.barh(y, t['cancel_rate'], color=BLUE, height=0.62)
    a2.set_title('Cancellation rate'); a2.xaxis.set_major_formatter(PCT)
    a2.set_xlim(0, max(t['cancel_rate'].max(), BASE) * 1.2); a2.grid(axis='y', visible=False)
    for yi, v in zip(y, t['cancel_rate']):
        a2.text(v, yi, f' {v:.1%}', va='center', fontsize=8.5, color=INK)
    a2.set_ylim(-0.6, len(t) - 0.4)
    baseline_v(a2, y=len(t) - 0.45)
    suptitle(fig, title, sub)
    plt.show()
    return t
```

- "Non Refund" bookings account to 99% cancellation rate
- Bookings with changes and special request are more committed


```python
# 2.3 Strength of association — Cramér's V for every candidate characteristic
def cramers_v(x, y):
    ct = pd.crosstab(x, y)
    chi2 = chi2_contingency(ct, correction=False)[0]
    n = ct.values.sum(); r, k = ct.shape
    return np.sqrt(chi2 / (n * (min(r, k) - 1)))

cands = {'Deposit type': 'deposit_type', 'Lead time band': 'lead_band', 'Market segment': 'market_segment',
         'Special requests': 'requests_band', 'Distribution channel': 'distribution_channel',
         'Customer type': 'customer_type', 'Booking changes': 'changes_band', 'Hotel type': 'hotel',
         'Previous cancellation': 'prev_cancel_flag', 'Length of stay': 'stay_band', 'Party composition': 'party',
         'Waiting list band': 'wait_band', 'Repeat guest': 'is_repeated_guest', 'Meal plan': 'meal',
         'Arrival month': 'arrival_mname', 'Has agent': 'has_agent',
         'Car parking requested*': 'required_car_parking_spaces', 'Country = PRT*': None}
rows = []
for label, col in cands.items():
    col = col or 'country_prt'
    x = df[col] if col != 'required_car_parking_spaces' else (df[col] > 0).astype(int)
    rows.append((label, cramers_v(x, df['is_canceled'])))
cv = pd.DataFrame(rows, columns=['feature', 'cramers_v']).sort_values('cramers_v')
fig, ax = plt.subplots(figsize=(9, 6.6))
cols = [CONTEXT if f.endswith('*') else BLUE for f in cv['feature']]
ax.barh(cv['feature'], cv['cramers_v'], color=cols, height=0.62)
for yi, v in enumerate(cv['cramers_v']):
    ax.text(v, yi, f' {v:.2f}', va='center', fontsize=8.5)
ax.set_xlim(0, cv['cramers_v'].max() * 1.15); ax.grid(axis='y', visible=False); ax.set_xlabel("Cramér's V (0 = none, 1 = perfect)")
suptitle(fig, '2.3  Deposit type, lead time and market segment are the strongest single signals',
         "Association with cancellation (Cramér's V). * grey = recorded at/after arrival, excluded from the model")
fig.subplots_adjust(left=0.25); plt.show()
```

Deposit type has the strongest unadjusted association with cancellation in this set (Cramér's V = 0.481).

```python
# 2.4 Non-refundable cancellations — which segments and channels do they come from?
nr = df[(df['is_canceled'] == 1) & (df['deposit_type'] == 'Non Refund')]
ct = pd.crosstab(nr['market_segment'], nr['distribution_channel'])
ct = ct.loc[ct.sum(axis=1).sort_values().index]                       # biggest segment at the top
CHANNEL_COLORS = {'TA/TO': BLUE, 'Direct': ORANGE, 'Corporate': AQUA, 'GDS': '#eda100'}   # fixed colour per channel
channels = [c for c in CHANNEL_COLORS if c in ct.columns]
ct = ct[channels]
total = ct.sum(axis=1)

fig, ax = plt.subplots(figsize=(10.5, 0.55 * len(ct) + 2.2))
left = np.zeros(len(ct))
for ch in channels:
    ax.barh(ct.index, ct[ch], left=left, color=CHANNEL_COLORS[ch], height=0.6, label=ch,
            edgecolor=SURFACE, linewidth=1.5)                           # thin gap between stacked pieces
    left += ct[ch].values
for y, (seg_name, t) in enumerate(total.items()):
    top_ch = ct.loc[seg_name].idxmax()
    ax.text(t, y, f'  {t:,}  ({ct.loc[seg_name, top_ch] / t:.0%} {top_ch})', va='center', fontsize=9, color=INK2)
ax.xaxis.set_major_formatter(THOU); ax.set_xlim(0, total.max() * 1.35); ax.grid(axis='y', visible=False)
ax.set_xlabel('Cancelled bookings with a Non Refund deposit')
ax.legend(title='Distribution channel', loc='lower right')
suptitle(fig, f'2.4  {len(nr):,} non-refundable cancellations — almost all Groups and Offline TA/TO via travel agents',
         'Cancelled bookings with deposit type = Non Refund: market segment (bars) split by distribution channel (colours)')
plt.show()
ct.assign(Total=total).sort_values('Total', ascending=False)
```

99% of the group cancellation are for bookings made via TA/TO and not by individual guest

```python
# 3.3 Logistic regression — odds ratio for each waiting-list band, unadjusted vs adjusted
import statsmodels.formula.api as smf
mdl = seg.copy()
mdl['wait_band'] = mdl['wait_band'].astype(str)
for c in ['lead_band', 'stay_band', 'party', 'requests_band', 'arrival_mname']:
    mdl[c] = mdl[c].astype(str)
ref = "C(wait_band, Treatment('0'))"
f_un = f"is_canceled ~ {ref}"
f_adj = (f"is_canceled ~ {ref} + C(hotel) + C(market_segment) + C(deposit_type) + C(customer_type) "
         "+ C(lead_band) + C(stay_band) + C(party) + C(requests_band) + C(arrival_mname) "
         "+ is_repeated_guest + prev_cancel_flag")
res = {}
for name, f in [('Unadjusted', f_un), ('Adjusted', f_adj)]:
    fit = smf.logit(f, data=mdl).fit(disp=0, maxiter=300)
    ci = fit.conf_int()
    keep = [i for i in fit.params.index if i.startswith(ref)]
    res[name] = pd.DataFrame({'OR': np.exp(fit.params[keep]).values, 'lo': np.exp(ci.loc[keep, 0]).values,
                              'hi': np.exp(ci.loc[keep, 1]).values, 'p': fit.pvalues[keep].values},
                             index=[k.split('[T.')[1].rstrip(']') for k in keep]).reindex(WAIT[1:])
fig, ax = plt.subplots(figsize=(9.5, 4))
y = np.arange(len(WAIT) - 1)
for k, (name, c) in enumerate([('Unadjusted', CONTEXT), ('Adjusted', BLUE)]):
    r = res[name]; yy = y + (0.15 if k == 0 else -0.15)
    ax.errorbar(r['OR'], yy, xerr=[r['OR'] - r['lo'], r['hi'] - r['OR']], fmt='o', color=c, ms=7,
                capsize=0, elinewidth=2, label=name, markeredgecolor=SURFACE)
ax.axvline(1, color=INK2, lw=1)
ax.set_xscale('log'); ax.xaxis.set_major_formatter(mtick.FuncFormatter(lambda v, _: f'{v:g}'))
ax.set_yticks(y, [f'{b} days' for b in WAIT[1:]]); ax.invert_yaxis(); ax.grid(axis='y', visible=False)
ax.set_xlabel('Odds ratio vs. no waiting (log scale, 95% CI)'); ax.legend(loc='lower right')
suptitle(fig, '3.3  Once booking profile is controlled for, waiting-list effects shrink and point in mixed directions',
         'Logistic regression: odds of cancellation by waiting-list band, before vs after controlling for booking profile')
plt.show()
pd.concat(res, axis=1).round(3)
```

Waiting list has no consistent effect on cancellation

### RO5 — High-volume, high-risk booking profiles

If the hotel only focus on a few types of customer, would cancellation rate
decrease?

NOTE:
- Profile is a combination of hotel + market segment + deposit type and lead-time band e.g., “City · Online TA · No Deposit · booked 6–12 months ahead”. The reason why i didnt include distribution channel as part of the profile its because it didnt move the needle for the share of all cancellations for the top 10 profile
- High-volume + high risk booking profile refer to bookings that have high cancellation rate and booking volume is substential

```python
pcols = ['hotel', 'market_segment', 'deposit_type', 'lead_band']
prof = seg.groupby(pcols, observed=True)['is_canceled'].agg(bookings='size', cancellations='sum').reset_index()
prof = prof[prof['bookings'] >= 300].copy()
prof['cancel_rate'] = prof['cancellations'] / prof['bookings']
prof['excess'] = prof['bookings'] * (prof['cancel_rate'] - BASE)
prof['profile'] = (prof['hotel'].str.replace(' Hotel', '') + ' · ' + prof['market_segment'] + ' · '
                   + prof['deposit_type'] + ' · ' + prof['lead_band'].astype(str) + 'd')
prof['priority'] = (prof['cancel_rate'] >= 0.5) & (prof['bookings'] >= 1000)
prof = prof.sort_values('excess', ascending=False).reset_index(drop=True)
top = prof[prof['priority']].head(10).copy(); top['rank'] = range(1, len(top) + 1)

# 5.1 Volume vs rate map
fig, ax = plt.subplots(figsize=(11, 6))
o = prof[~prof['priority']]; p = prof[prof['priority']]
ax.scatter(o['bookings'], o['cancel_rate'], s=45, color=CONTEXT, edgecolor=SURFACE, linewidth=1.5, label='Other profiles (≥300 bookings)')
ax.scatter(p['bookings'], p['cancel_rate'], s=70, color=BLUE, edgecolor=SURFACE, linewidth=2, label='Priority: ≥50% rate and ≥1,000 bookings')
for _, r in top.iterrows():
    ax.annotate(str(r['rank']), (r['bookings'], r['cancel_rate']), xytext=(6, 4), textcoords='offset points',
                fontsize=9, fontweight='bold')
ax.axhline(0.5, color=INK2, lw=1, ls=(0, (4, 3))); ax.axvline(1000, color=INK2, lw=1, ls=(0, (4, 3)))
ax.set_xscale('log'); ax.xaxis.set_major_formatter(THOU); ax.yaxis.set_major_formatter(PCT); ax.set_ylim(0, 1.05)
ax.set_xlabel('Bookings in profile (log scale)'); ax.set_ylabel('Cancellation rate'); ax.legend(loc='lower left', bbox_to_anchor=(0, 1.0), ncol=2)
suptitle(fig, f'5.1  {len(p)} profiles combine high risk with high volume',
         'Each dot = hotel × segment × deposit × lead-time profile; numbers match ranks in 5.2', gap=1.3)
plt.show()
```

```python
# 5.2 Ranked priority profiles by excess cancellations
t = top.iloc[::-1]
fig, ax = plt.subplots(figsize=(11, 0.48 * len(t) + 1.8))
ax.barh(range(len(t)), t['excess'], color=BLUE, height=0.62)
ax.set_yticks(range(len(t)), [f"{r}. {p}" for r, p in zip(t['rank'], t['profile'])])
for yi, (_, r) in enumerate(t.iterrows()):
    ax.text(r['excess'], yi, f"  {r['excess']:,.0f}   (n={r['bookings']:,}, {r['cancel_rate']:.0%})", va='center', fontsize=8.5, color=INK2)
ax.set_xlim(0, t['excess'].max() * 1.45); ax.xaxis.set_major_formatter(THOU); ax.grid(axis='y', visible=False)
ax.set_xlabel('Excess cancellations vs. overall rate')
suptitle(fig, '5.2  Top priority profiles ranked by excess cancellations', 'Cancellations above the 37% baseline that each profile generates')
fig.subplots_adjust(left=0.38); plt.show()
top[['rank', 'profile', 'bookings', 'cancel_rate', 'excess']].round(3)
```
